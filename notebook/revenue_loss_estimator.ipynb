```python
import pandas as pd
df=pd.read_csv("../data/customer_retention_result.csv")
```

# Revenue at risk for each customer

```python

df["Revenue_At_Risk"]=(df["MonthlyCharges"]*df["Churn_Probability"])
```

# total expected monthly revenue loss

```python

total_loss=df['Revenue_At_Risk'].sum()
print(f"Total Expected Monthly Revenue Loss:${total_loss}")
```

# find the top 10 customers by revenue at risk:

```python

top_10=df.sort_values(by='Revenue_At_Risk',ascending=False).head(10)
print(top_10[
    [
        "MonthlyCharges",
        "Churn_Probability",
        "Revenue_At_Risk",
        "Retention_Action"
    ]
])
```

# Save Revenue Loss Estimator results

```python

df.to_csv("revenue_loss_estimator.csv", index=False)

print("Revenue Loss Estimator saved successfully!")
df
```
